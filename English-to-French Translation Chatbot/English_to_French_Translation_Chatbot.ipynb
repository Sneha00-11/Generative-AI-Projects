# 🇬🇧➡️🇫🇷 English-to-French Translation Chatbot

This notebook builds an interactive chatbot that translates English text into French using a pre-trained Hugging Face translation model (`Helsinki-NLP/opus-mt-en-fr`). The chatbot simulates a natural, real-time typing effect when displaying the translated response.

**What this notebook covers:**
1. Installing and importing required libraries
2. Loading a pre-trained Hugging Face translation model
3. Writing a translation function
4. Adding a dynamic typing effect for the chatbot's replies
5. Running an interactive chatbot loop

> Tip: Run the cells in order, from top to bottom.

## 1. Install Required Libraries

We need `transformers` (to load the pre-trained translation model), `sentencepiece` (the tokenizer used by the Helsinki-NLP MarianMT models), and `torch` (the deep learning backend).

```python
# Install the libraries needed to run the Hugging Face translation model.
# transformers -> gives us access to pre-trained NLP models and pipelines
# sentencepiece -> tokenizer required by the MarianMT (Helsinki-NLP) models
# torch -> the underlying deep learning framework the model runs on
!pip install -q transformers sentencepiece torch
```

## 2. Import Libraries

- `pipeline` from `transformers` gives us a simple, high-level interface for running the translation model.
- `time` and `sys` are used to build the dynamic typing effect.
- `textwrap` is used to keep printed output tidy.

```python
from transformers import pipeline
import time
import sys
import textwrap
```

## 3. Load the Pre-trained Translation Model

We use `Helsinki-NLP/opus-mt-en-fr`, a MarianMT model trained specifically for English-to-French translation and hosted on the Hugging Face Hub. Loading it through the `pipeline` API automatically handles downloading the model weights, the tokenizer, and setting up the translation task.

```python
# Load the English -> French translation pipeline.
# task='translation_en_to_fr' tells Hugging Face what kind of pipeline to build.
# model='Helsinki-NLP/opus-mt-en-fr' specifies the exact pre-trained model to use.
# This may take a minute the first time, since the model weights (~300MB) are downloaded.
print("Loading translation model... please wait.")
translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")
print("Model loaded successfully!")
```

## 4. Translation Function

This function takes an English sentence, passes it to the pipeline, and returns the translated French text. We wrap the pipeline call so the rest of our chatbot code stays clean and reusable.

```python
def translate_to_french(text):
    """
    Translate a piece of English text into French using the Hugging Face
    translation pipeline.

    Parameters:
        text (str): The English sentence to translate.

    Returns:
        str: The translated French sentence.
    """
    # The pipeline returns a list of dictionaries, e.g. [{'translation_text': 'Bonjour le monde !'}]
    result = translator(text)
    return result[0]["translation_text"]


# Quick sanity check
sample = "Hello, world!"
print(f"English: {sample}")
print(f"French:  {translate_to_french(sample)}")
```

## 5. Dynamic Typing Effect

To make the chatbot feel more natural and engaging, we print the French translation one character at a time with a small delay, instead of dumping the whole sentence on screen instantly.

```python
def type_effect(text, delay=0.03, prefix="Chatbot: "):
    """
    Print text character-by-character to simulate a natural typing effect.

    Parameters:
        text (str): The text to display.
        delay (float): Time (in seconds) to wait between each character.
        prefix (str): A label printed before the typed text (e.g. 'Chatbot: ').
    """
    sys.stdout.write(prefix)
    sys.stdout.flush()
    for char in text:
        sys.stdout.write(char)
        sys.stdout.flush()
        time.sleep(delay)
    print()  # move to a new line once typing is finished
```

## 6. Chatbot Loop

This is the main interactive loop. The user types an English sentence, the model translates it into French, and the chatbot 'types' the answer back in real time. Typing `exit`, `quit`, or `bye` ends the conversation.

```python
def run_chatbot():
    """
    Main chatbot loop.
    Continuously accepts English input from the user, translates it to
    French, and displays the result with a typing animation until the
    user chooses to exit.
    """
    exit_words = {"exit", "quit", "bye"}

    print("=" * 55)
    print(" English -> French Translation Chatbot")
    print(" Type a sentence in English, or 'exit' to stop.")
    print("=" * 55)

    while True:
        user_input = input("\nYou: ").strip()

        if not user_input:
            # Skip empty input instead of sending it to the model
            print("Chatbot: Please type something in English!")
            continue

        if user_input.lower() in exit_words:
            type_effect("Au revoir! (Goodbye!)")
            break

        # Translate the input text using our Hugging Face model
        french_translation = translate_to_french(user_input)

        # Display the translation with the typing effect
        type_effect(french_translation)


# Run the chatbot (uncomment the line below in Colab / Jupyter to start chatting)
# run_chatbot()
```

## 7. Try It Out

Run the cell below to start chatting. Enter English phrases and watch them appear in French with a live typing effect. Type `exit`, `quit`, or `bye` to end the session.

```python
run_chatbot()
```

## 8. Batch Demo (Optional)

A quick, non-interactive demonstration translating several example phrases, useful for showing off the chatbot's accuracy without needing manual input.

```python
demo_phrases = [
    "Hello, world!",
    "How are you today?",
    "I love learning about artificial intelligence.",
    "Where is the nearest train station?",
    "Thank you very much for your help."
]

for phrase in demo_phrases:
    translation = translate_to_french(phrase)
    print(f"EN: {phrase}\nFR: {translation}\n" + "-" * 40)
```
